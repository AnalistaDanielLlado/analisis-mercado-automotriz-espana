```python
#Matriculacion por tipo de vehiculo
matr_tipo = pd.read_excel(
    file_matric,
    sheet_name="Matr_tipo_veh",
    skiprows=1
)

# Eliminar filas vacías
matr_tipo = matr_tipo.dropna()

# Renombrar columnas
matr_tipo.columns = [
    "anio",
    "camiones_furgonetas",
    "autobuses",
    "turismos",
    "motocicletas",
    "tractores_industriales",
    "otros_vehiculos",
    "total"
]

# Eliminar la fila que contiene el texto "Años"
matr_tipo = matr_tipo[matr_tipo["anio"] != "Años"]

# Convertir a números
matr_tipo["anio"] = matr_tipo["anio"].astype(int)
matr_tipo["total"] = matr_tipo["total"].astype(int)

matr_tipo.head()
```

```python
#distintivos medioambientales
m_tur_medio = pd.read_excel(
    file_tablas,
    sheet_name="M_TUR_MEDIO",
    skiprows=1
)

# Eliminar filas vacías
m_tur_medio = m_tur_medio.dropna()

# Usar la primera fila como nombres de columnas
m_tur_medio.columns = m_tur_medio.iloc[0]

# Eliminar esa fila
m_tur_medio = m_tur_medio[1:]

# Renombrar columna Provincias
m_tur_medio = m_tur_medio.rename(columns={"Provincias": "provincia"})

m_tur_medio.head()
```

```python
m_tur_medio.info()
```

```python
#convierto a numerico
cols = ["CERO", "ECO", "C", "B", "Sin distintivo"]

for col in cols:
    m_tur_medio[col] = pd.to_numeric(m_tur_medio[col], errors="coerce")
```

```python
#creo totales y porcentajes
m_tur_medio["total"] = (
    m_tur_medio["CERO"] +
    m_tur_medio["ECO"] +
    m_tur_medio["C"] +
    m_tur_medio["B"] +
    m_tur_medio["Sin distintivo"]
)

m_tur_medio["pct_cero"] = m_tur_medio["CERO"] / m_tur_medio["total"]
m_tur_medio["pct_eco"] = m_tur_medio["ECO"] / m_tur_medio["total"]
m_tur_medio["pct_c"] = m_tur_medio["C"] / m_tur_medio["total"]
m_tur_medio["pct_b"] = m_tur_medio["B"] / m_tur_medio["total"]
m_tur_medio["pct_sin"] = m_tur_medio["Sin distintivo"] / m_tur_medio["total"]
```

```python
#guardo
m_tur_medio.to_csv(
    PROCESSED_DIR / "distintivo_ambiental.csv",
    index=False
)
```

```python
from pathlib import Path
import pandas as pd

PROCESSED_DIR = Path("/mnt/c/Users/danie/OneDrive/Desktop/Proyectos/Automocion/data/processed")

matr_gasolina = pd.read_csv(PROCESSED_DIR / "matriculaciones_gasolina.csv")
matr_gasoil = pd.read_csv(PROCESSED_DIR / "matriculaciones_gasoil.csv")
matr_otros = pd.read_csv(PROCESSED_DIR / "matriculaciones_otros.csv")

combustible = matr_gasolina[["anio", "total"]].merge(
    matr_gasoil[["anio", "total"]],
    on="anio",
    suffixes=("_gasolina", "_gasoil")
)

combustible = combustible.merge(
    matr_otros[["anio", "total"]],
    on="anio",
    how="left"
)

combustible = combustible.rename(columns={"total": "otros"})
combustible = combustible.fillna(0)

combustible["total"] = (
    combustible["total_gasolina"] +
    combustible["total_gasoil"] +
    combustible["otros"]
)

combustible = combustible.rename(columns={
    "total_gasolina": "gasolina",
    "total_gasoil": "diesel"
})

combustible.head()
```

```python
combustible.to_csv(PROCESSED_DIR / "combustible.csv", index=False)
```
